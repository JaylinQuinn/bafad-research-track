```python
import numpy as np
import matplotlib.pyplot as plt

# Reproducible synthetic sensor stream (3 sensors, 200 time steps)
np.random.seed(42)
t = np.linspace(0, 10, 200)
sensor_stream = np.stack(
    [np.sin(t * (i + 1)) + np.random.normal(0, 0.05, len(t)) for i in range(3)],
    axis=1
)

# Verify shape and value range
print(f"Sensor stream shape : {sensor_stream.shape}")
print(f"Value range         : [{sensor_stream.min():.3f}, {sensor_stream.max():.3f}]")
assert sensor_stream.shape == (200, 3), "Shape mismatch — check stack call"

# Plot all 3 sensors
fig, axes = plt.subplots(3, 1, figsize=(10, 6), sharex=True)
colors = ['#0EA5E9', '#10B981', '#F59E0B']

for i, (ax, color) in enumerate(zip(axes, colors)):
    ax.plot(t, sensor_stream[:, i], color=color, linewidth=1.2, label=f'Sensor {i+1}')
    ax.set_ylabel(f'S{i+1}', fontsize=10)
    ax.legend(loc='upper right', fontsize=9)
    ax.set_ylim(-1.5, 1.5)

axes[-1].set_xlabel('Time (arbitrary units)', fontsize=10)
fig.suptitle('BAFAD — Sensor Array Sanity Check (Week 1)', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('week01_sanity_check.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n✓ Sanity check passed — environment ready for BAFAD research.")
```
